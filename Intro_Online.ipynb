```python
# Chequear palindromo

# ABBA  = ABBA
# APPLE != ELPPA

def check_palindrome(word):

  reversed_word = word[::-1]

  return reversed_word == word
```

```python
check_palindrome("APPLE")
```

```python
# Dada una lista construir una lista de tuplas con cada numero y su cuadrado

list_numbers = [1,2,4,5,6,7,2,5]

list_tuples = [(x, x**2) for x in list_numbers]
list_tuples
```

```python
# Diccionario con clave el numero y valor su cuadrado

dict_cuadrados = {x: x**2 for x in list_numbers}
dict_cuadrados
```

```python
# Lista de numeros -> Diccionario con clave el numero y valor n_veces que se repite

dict_output = {x: 0 for x in list_numbers}

for x in list_numbers:
  dict_output[x] += 1

dict_output
```

```python
dict_output = {x: list_numbers.count(x) for x in list_numbers}
dict_output
```

```python
# SOLUCION RECURSIVA FIBONACCI

def fibonacci(n):
  if n == 0 or n == 1:
    return n
  return fibonacci(n-1) + fibonacci(n-2)
```

```python
fibonacci(4)
```

```python
%%timeit

fibonacci(10)
```

```python
from functools import lru_cache

@lru_cache
def fibonacci_cache(n):

  if n == 0 or n == 1:
    return n

  return fibonacci(n-1) + fibonacci(n-2)
```

```python
%%timeit

fibonacci_cache(10)
```

```python
# Dado un numeor binario como texto calcular el maximo numero de 0s y 1s consecutivos


n = "10010001110"

consecutive0 = max([len(x) for x in n.split("1")])
consecutive1 = max([len(x) for x in n.split("0")])
```

```python
# Añadir separador de miles a un numero N

import math

n = 10425231

sep = math.floor(len(str(n))/3) + 1
output_number = ""
for i in range(sep):
  group3 = str(n)[::-1][3*i:3*(i+1)]
  output_number = output_number + group3 + ","

final_number = output_number[::-1][1:]
final_number
```
