```python
# PROBLEM 7
import numpy as np

employee["age_on_hire_date"] = pd.to_datetime(employee['hire_date']) - pd.to_datetime(employee['birth_date'])
employee["age_on_hire_date"] = np.floor(employee["age_on_hire_date"].dt.days/365)
employee
```
